# Opening and Reading Files
Python can open a file so your program can read its contents. This chapter explains how to open and close files and read all or part of a file.

## Open a File
Use `open(path, mode)` to open a file. The mode `"r"` means read, and it is the default. The path can be a file name or a path to a file. This example creates a sample file in a temporary folder, so it does not depend on files on your computer.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as folder:
    path = Path(folder) / "notes.txt"
    path.write_text("These are the file contents.", encoding="utf-8")

    file = open(path, "r", encoding="utf-8")
    print(file.read())
    file.close()
```

## Using the `with` Statement
The `with` statement closes the file automatically when the indented block ends. It is the recommended way to work with files.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "greeting.txt"
    path.write_text("Hello!", encoding="utf-8")

    with open(path, encoding="utf-8") as file:
        print(file.read())
```

## Close Files
If you open a file without `with`, call its `close()` method when you finish. Closing releases the file resource. Prefer `with` because it closes the file even if an error happens.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "message.txt"
    path.write_text("Read this message.", encoding="utf-8")

    file = open(path, encoding="utf-8")
    try:
        print(file.read())
    finally:
        file.close()

    print("File is closed:", file.closed)
```

## Read Only Parts of a File
Pass a number to `read()` to read up to that many characters. The next call continues from where the previous call stopped.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "sentence.txt"
    path.write_text("Python files are useful.", encoding="utf-8")

    with open(path, encoding="utf-8") as file:
        print(file.read(6))
        print(file.read(5))
```

## Read Lines
Use `readline()` to read one line at a time. You can also loop through the file to read every line. Lines usually include a newline character, so `strip()` can remove it for display.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "shopping.txt"
    path.write_text("bread\nmilk\napples\n", encoding="utf-8")

    with open(path, encoding="utf-8") as file:
        print("First line:", file.readline().strip())
        print("Other lines:")
        for line in file:
            print(line.strip())
```
