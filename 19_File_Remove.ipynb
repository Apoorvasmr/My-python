# Deleting Files and Folders
Python can remove files and empty folders. Deleting is permanent, so check the path carefully. The examples in this chapter only create and delete items inside temporary folders.

## Delete a File
Use `unlink()` to delete a file. It raises an error if the file does not exist, so check the path first when it may be missing.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as folder:
    path = Path(folder) / "sample.txt"
    path.write_text("Temporary file", encoding="utf-8")

    path.unlink()
    print("File exists:", path.exists())
```

## Check if a File Exists
Use `exists()` before deleting a file when you are not sure it is there. This can prevent a missing-file error.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "sample.txt"
    path.write_text("Temporary file", encoding="utf-8")

    if path.exists():
        path.unlink()
        print("File deleted.")
    else:
        print("File does not exist.")
```

## Delete a Folder
Use `rmdir()` to delete an empty folder. It will not delete a folder that still contains files. Check that the folder is the one you intend to remove.

```python
with TemporaryDirectory() as folder:
    empty_folder = Path(folder) / "empty_folder"
    empty_folder.mkdir()

    empty_folder.rmdir()
    print("Folder exists:", empty_folder.exists())
```
