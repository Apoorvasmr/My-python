# Python Try and Except
An exception is an error that happens while a program is running. Use `try` and `except` to handle errors that you expect, so the program can respond clearly.

## Exception Handling
Put code that might cause an expected error inside `try`. If that error happens, Python runs the matching `except` block.

```python
try:
    number = int("hello")
except ValueError:
    print("Please use a valid whole number.")
```

Catch the kind of error you expect. For example, converting invalid text to a number raises `ValueError`, while dividing by zero raises `ZeroDivisionError`.

```python
try:
    answer = 10 / 0
except ZeroDivisionError:
    print("A number cannot be divided by zero.")
```

## Many Exceptions
You can use more than one `except` block to handle different errors in different ways. Python runs the first block that matches the error.

```python
value = "0"

try:
    number = int(value)
    result = 10 / number
except ValueError:
    print("The value must be a whole number.")
except ZeroDivisionError:
    print("The number must not be zero.")
else:
    print(result)
```

You can use `as` to give the error a name and show its message. Only display error details when they are useful to the person using your program.

```python
try:
    int("not a number")
except ValueError as error:
    print("Could not convert the value:", error)
```

## `else`
The `else` block runs only when the code in `try` finishes without an exception. Use it for code that should run after a successful attempt.

```python
try:
    number = int("12")
except ValueError:
    print("That was not a whole number.")
else:
    print("The number is", number)
```

## `finally`
The `finally` block runs whether an exception happens or not. It is often used for cleanup work.

```python
try:
    int("not a number")
except ValueError:
    print("The value could not be converted.")
finally:
    print("This message always appears.")
```

## Raise an Exception
Use `raise` to report an error when your code receives a value it cannot accept. Choose an exception type that describes the problem.

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    return age

try:
    check_age(-1)
except ValueError as error:
    print(error)
```
