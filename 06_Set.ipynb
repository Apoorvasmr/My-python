## Python Sets
A set is a collection that stores multiple items in one variable. Sets are unordered, changeable, and do not allow duplicate values. They are useful when you need to store unique items and check membership quickly.



```python
fruits = {"apple", "banana", "cherry"}
print(fruits)
```

## Sets are unordered
A set does not keep items in a fixed order. Because of that, you cannot access items by index like you can with a list or tuple.



```python
numbers = {5, 2, 7, 1}
print(numbers)
print(numbers)
```

## Duplicate values are ignored
If you add the same item more than once to a set, it only appears once. Python removes duplicates automatically.



```python
items = {"apple", "banana", "cherry", "apple", "banana"}
print(items)
print(len(items))
```

## Set length
Use the built-in `len()` function to count how many unique items a set contains.


```python
numbers = {1, 2, 3, 4}
print(len(numbers))
```

## Set type
Use the built-in `type()` function to check an object's data type. A set has the type `set`.


```python
numbers = {1, 2, 3, 4}
print(type(numbers))
```

## Accessing set items
You cannot access set items by index, but you can loop through the set or check whether a value is inside it with `in`.



```python
fruits = {"apple", "banana", "cherry"}
for x in fruits:
    print(x)

print("banana" in fruits)
print("mango" in fruits)
```

## add()
Use the `add()` method to insert one item into a set. If the value already exists, the set stays unchanged.


```python
fruits = {"apple", "banana"}
fruits.add("cherry")
print(fruits)
```

## update()
Use the `update()` method to add every item from another iterable, such as a list or another set, to the set.


```python
fruits = {"apple", "banana"}
more_fruits = ["orange", "mango"]
fruits.update(more_fruits)
print(fruits)
```

## remove()
The `remove()` method deletes a specified item from a set. It raises a `KeyError` if the item is not present.


```python
fruits = {"apple", "banana", "cherry", "date"}
fruits.remove("banana")
print(fruits)
```

## discard()
The `discard()` method removes a specified item if it exists. Unlike `remove()`, it does not raise an error if the item is missing.


```python
fruits = {"apple", "banana", "cherry"}
fruits.discard("banana")
fruits.discard("mango")
print(fruits)
```

## pop()
The `pop()` method removes and returns an arbitrary item from a set. Because sets are unordered, you cannot choose which item it removes.


```python
fruits = {"apple", "banana", "cherry"}
removed_item = fruits.pop()
print("Removed item:", removed_item)
print("Remaining items:", fruits)
```

## clear()
The `clear()` method removes all items from the set but keeps the set variable, leaving it empty.


```python
fruits = {"apple", "banana", "cherry"}
fruits.clear()
print(fruits)
```

## del keyword
The `del` keyword deletes the set variable itself. After deletion, that variable name can no longer be used.


```python
fruits = {"apple", "banana", "cherry"}
del fruits
print("The set variable has been deleted.")
```

## What are set operations?
Set operations compare or combine sets based on which items they share. Union combines unique items, intersection finds shared items, difference finds items in one set but not another, and symmetric difference finds items that are in either set but not both. These operations return results as sets.


```python
group_a = {"red", "blue", "green"}
group_b = {"blue", "yellow"}

print("All unique colors:", group_a | group_b)
print("Shared colors:", group_a & group_b)
print("Only in group_a:", group_a - group_b)
print("In either, but not both:", group_a ^ group_b)
```

## What are set methods?
Set methods are built-in functions called on a set with dot notation, such as `set_a.union(set_b)`. Some return a new set, some return `True` or `False`, and update methods change the original set. Many set operations also have operator shortcuts, such as `|` for union.


```python
set_a = {"red", "blue"}
set_b = {"blue", "yellow"}

combined = set_a.union(set_b)
print("Union result:", combined)
print("Original set_a:", set_a)

set_a.update({"green"})
print("set_a after update():", set_a)
print("No shared items with black:", set_a.isdisjoint({"black"}))
```

## union()
The `union()` method combines the items from two or more sets and returns a new set containing each unique item. The `|` operator also performs a union.


```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(set_a.union(set_b))
print(set_a | set_b)
```

## intersection()
The `intersection()` method returns a new set containing only items shared by both sets. The `&` operator also performs an intersection.


```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(set_a.intersection(set_b))
print(set_a & set_b)
```

## intersection_update()
The `intersection_update()` method keeps only items shared with another set. It changes the original set and does not return a new result set.


```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
set_a.intersection_update(set_b)
print(set_a)
```

## difference()
The `difference()` method returns a new set of items that are in the first set but not in the second. The `-` operator also performs this operation.


```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(set_a.difference(set_b))
print(set_a - set_b)
```

## difference_update()
The `difference_update()` method removes items from the original set when they also appear in another set. Unlike `difference()`, it changes the set in place.


```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
set_a.difference_update(set_b)
print(set_a)
```

## symmetric_difference()
The `symmetric_difference()` method returns a new set containing items that appear in either set, but not in both. The `^` operator does the same operation.


```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print(set_a.symmetric_difference(set_b))
print(set_a ^ set_b)
```

## symmetric_difference_update()
The `symmetric_difference_update()` method replaces the original set's contents with items that are in either set but not both.


```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
set_a.symmetric_difference_update(set_b)
print(set_a)
```

## isdisjoint()
The `isdisjoint()` method returns `True` if two sets have no items in common. It returns `False` if they share at least one item.


```python
set_a = {1, 2, 3}
set_b = {4, 5, 6}
set_c = {3, 7, 8}
print(set_a.isdisjoint(set_b))
print(set_a.isdisjoint(set_c))
```

## issubset()
The `issubset()` method returns `True` when every item in one set is also present in another set.


```python
set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}
print(set_a.issubset(set_b))

set_c = {1, 2, 6}
print(set_c.issubset(set_b))
```

## issuperset()
The `issuperset()` method returns `True` when a set contains every item from another set.


```python
set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}
print(set_b.issuperset(set_a))

set_c = {1, 2, 6}
print(set_b.issuperset(set_c))
```

## copy()
The `copy()` method creates a separate copy of a set, so adding items to the copy does not change the original set.


```python
numbers = {1, 2, 3, 4}

numbers_copy = numbers.copy()

numbers_copy.add(5)

print("Original set:", numbers)
print("Copied set:", numbers_copy)
```

## frozenset()
A `frozenset` is an immutable version of a set. You can iterate over its items and use set operations, but you cannot add or remove items after it is created. Create one with `frozenset()` using an iterable.


```python
numbers = frozenset([1, 2, 3, 2])
print(numbers)
print(2 in numbers)
print(numbers | frozenset([3, 4]))
```
