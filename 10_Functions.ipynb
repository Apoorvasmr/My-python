# Python Functions
A function is a named group of instructions. You can run it whenever you need those instructions.

Functions help organize programs and avoid repeating the same code.

## Creating a Function
Use `def`, followed by the function name and parentheses. Indent the instructions inside the function.

```python
def say_hello():
    print("Hello!")
```

## Calling a Function
A function runs when you call it by writing its name followed by parentheses.

```python
say_hello()
```

## Function Names
Choose a name that describes what the function does. Use letters, numbers, and underscores, but do not start a name with a number. Python treats uppercase and lowercase letters as different.

```python
def show_message():
    print("Learning Python")

show_message()
```

## Why Use Functions?
A function lets you reuse instructions. If the task needs to change, you can update the function in one place.

```python
def greet(name):
    print("Hello,", name)

greet("Asha")
greet("Leo")
```

## Arguments
Arguments are values you give a function when you call it.

### Parameters vs Arguments
A parameter is a name in the function definition. An argument is the value supplied when the function is called.

```python
def greet(name):  # name is a parameter
    print("Hello,", name)

greet("Mina")  # "Mina" is an argument
```

## Number of Arguments
A function must receive the required number of arguments. If a function has two required parameters, call it with two values.

```python
def add_numbers(first, second):
    print(first + second)

add_numbers(4, 6)
```

## Positional Arguments
Positional arguments are matched to parameters by their order.

```python
def describe_pet(name, animal):
    print(name, "is a", animal)

describe_pet("Pip", "cat")
```

## Keyword Arguments
Keyword arguments use parameter names. You can give them in a different order from the function definition.

```python
def describe_pet(name, animal):
    print(name, "is a", animal)

describe_pet(animal="dog", name="Rex")
```

## Mixing Positional and Keyword Arguments
You can use both kinds in one call. Put positional arguments first, then keyword arguments.

```python
def describe_pet(name, animal):
    print(name, "is a", animal)

describe_pet("Milo", animal="cat")
```

## Default Parameter Values
A parameter can have a default value. Python uses it when the call does not provide that argument.

```python
def greet(name="friend"):
    print("Hello,", name)

greet()
greet("Sam")
```

## Passing Different Data Types
A function can receive values of different types, such as strings, numbers, and lists. The function can work with the value it receives.

```python
def show_items(items):
    for item in items:
        print(item)

show_items(["pen", "book", "bag"])
```

## Return Values
Use `return` to send a result back to the code that called the function. The caller can save or use that result.

```python
def square(number):
    return number * number

answer = square(5)
print(answer)
```

## Returning Different Data Types
A function can return a number, string, list, or another type of value.

```python
def make_profile():
    return {"name": "Ari", "age": 12}

profile = make_profile()
print(profile["name"])
```

## The `pass` Statement
Use `pass` when you need a function body but have not written its instructions yet. It keeps the code valid.

```python
def todo_later():
    pass
```

## Positional-Only Arguments
Parameters before `/` must be passed by position, not by name.

```python
def divide(number, divisor, /):
    return number / divisor

print(divide(10, 2))
```

## Keyword-Only Arguments
Parameters after `*` must be passed using their names.

```python
def create_account(name, *, active):
    print(name, active)

create_account("Jo", active=True)
```

## Combining Positional-Only and Keyword-Only Arguments
A function can use `/` and `*` to control how its parameters are passed.

```python
def send_message(text, /, *, urgent=False):
    print(text, "(urgent:", urgent, ")")

send_message("Call home", urgent=True)
```

## Arbitrary Positional Arguments: `*args`
### What is `*args`?
Use `*args` when a function should accept any number of extra positional arguments. Inside the function, `args` is a tuple.

```python
def add_all(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(add_all(2, 4, 6))
```

### Using `*args` with Regular Arguments
You can use regular parameters before `*args`. The regular parameters receive the first values; `args` collects the remaining values.

```python
def show_team(captain, *players):
    print("Captain:", captain)
    print("Other players:", players)

show_team("Nia", "Max", "Lee")
```

### Practical Example with `*args`
A function can use `*args` to work with any number of scores.

```python
def average(*scores):
    if not scores:
        return 0
    return sum(scores) / len(scores)

print(average(80, 90, 100))
```

## Arbitrary Keyword Arguments: `**kwargs`
### What is `**kwargs`?
Use `**kwargs` when a function should accept any number of named arguments. Inside the function, `kwargs` is a dictionary.

```python
def show_details(**details):
    for key, value in details.items():
        print(key, "=", value)

show_details(name="Kai", city="Delhi")
```

### Using `**kwargs` with Regular Arguments
You can use regular parameters before `**kwargs`. The named values not matched to those parameters are collected in a dictionary.

```python
def show_person(name, **details):
    print("Name:", name)
    print("Other details:", details)

show_person("Kai", city="Delhi", age=20)
```

### Combining `*args` and `**kwargs`
You can combine regular parameters, `*args`, and `**kwargs`. Keep them in this order in the function definition.

```python
def show_order(order_id, *items, **details):
    print("Order:", order_id)
    print("Items:", items)
    print("Details:", details)

show_order(101, "tea", "bread", delivery="home")
```

## Unpacking Arguments
Use `*` to unpack a list or tuple into positional arguments. Use `**` to unpack a dictionary into keyword arguments.

### Unpacking Lists with `*`
The `*` operator sends each list item to a separate positional parameter.

```python
def show_pair(first, second):
    print(first, second)

colors = ["red", "blue"]
show_pair(*colors)
```

### Unpacking Dictionaries with `**`
The `**` operator sends dictionary values to parameters with matching names.

```python
def show_pair(first, second):
    print(first, second)

sizes = {"first": "small", "second": "large"}
show_pair(**sizes)
```

## Recursion
Recursion happens when a function calls itself.

### Base Case and Recursive Case
The base case stops the calls. The recursive case calls the function again and moves toward the base case.

```python
def count_down(number):
    if number <= 0:  # base case
        print("Done")
        return
    print(number)
    count_down(number - 1)  # recursive case

count_down(3)
```

A recursive function can also return values. Here, each call multiplies the number by the result of the next smaller call.

```python
def factorial(number):
    if number <= 1:
        return 1
    return number * factorial(number - 1)

print(factorial(5))
```

## Fibonacci Sequence with Recursion
Each Fibonacci number is the sum of the two before it. This example uses recursion to calculate the number at a position.

```python
def fibonacci(position):
    if position <= 1:
        return position
    return fibonacci(position - 1) + fibonacci(position - 2)

print([fibonacci(position) for position in range(7)])
```

## Recursion with Lists
A function can process a list recursively by handling one item and then calling itself with the remaining items.

```python
def sum_list(items):
    if not items:
        return 0
    return items[0] + sum_list(items[1:])

print(sum_list([2, 3, 4]))
```

## Recursion Depth Limit
Python limits how many function calls can be active at once. A recursive function without a correct base case may keep calling itself until Python raises a `RecursionError`. Use a loop when it makes the task simpler.

```python
# This function stops because it moves toward the base case.
def safe_count(number):
    if number == 0:
        return
    safe_count(number - 1)

safe_count(5)
```

## Variable Scope
Scope is the part of a program where a variable name can be used. A variable created inside a function is usually available only inside that function.

### Local Scope
A variable created inside a function has local scope. It can be used inside that function.

```python
def show_color():
    color = "green"
    print(color)

show_color()
```

### Function Inside a Function
A function can be defined inside another function. The inner function can use variables from the outer function.

```python
def outer():
    message = "Hello"

    def inner():
        print(message)

    inner()

outer()
```

### Global Scope
A variable created outside all functions has global scope. Functions can read it.

```python
greeting = "Welcome"

def show_greeting():
    print(greeting)

show_greeting()
```

### Naming Variables
A local variable can have the same name as a global variable. The local name is used inside the function, while the global value stays unchanged.

```python
message = "Outside"

def show_message():
    message = "Inside"
    print(message)

show_message()
print(message)
```

### The `global` Keyword
Use `global` inside a function when you want to assign a new value to a global variable.

```python
visits = 0

def add_visit():
    global visits
    visits += 1

add_visit()
print(visits)
```

### The `nonlocal` Keyword
Use `nonlocal` in a nested function to change a variable from the nearest outer function.

```python
def outer_counter():
    count = 0

    def add_one():
        nonlocal count
        count += 1

    add_one()
    print(count)

outer_counter()
```

### The LEGB Rule
When Python looks up a name, it checks these scopes in order: Local, Enclosing (outer functions), Global, and Built-in.

```python
word = "global"

def check_name():
    word = "local"
    print(word)  # Local value is found first.
    print(len(word))  # len is a built-in name.

check_name()
```

## Lambda Functions
A lambda is a small function written in one expression. It can take any number of arguments but has one expression as its result.

### Lambda Syntax
Write `lambda`, then its parameters, a colon, and one expression.

```python
double = lambda number: number * 2
print(double(4))
```

### Why Use Lambda Functions?
Lambdas are useful for short functions that are needed briefly, often as an argument to another function. Use `def` for longer or reusable functions.

```python
multiply = lambda first, second: first * second
print(multiply(3, 5))
```

### Lambda with Built-in Functions
A lambda can provide a short rule to built-in functions such as `map()`, `filter()`, and `sorted()`.

#### Using Lambda with `map()`
`map()` applies a function to each item. Use `list()` to see all the results.

```python
numbers = [1, 2, 3]
doubled = list(map(lambda number: number * 2, numbers))
print(doubled)
```

#### Using Lambda with `filter()`
`filter()` keeps items for which the function returns `True`.

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda number: number % 2 == 0, numbers))
print(even_numbers)
```

#### Using Lambda with `sorted()`
The `key` argument tells `sorted()` what value to use for comparing items.

```python
words = ["pear", "fig", "apple"]
by_length = sorted(words, key=lambda word: len(word))
print(by_length)
```

## Decorators
A decorator is a function that adds behavior to another function. The `@` syntax applies a decorator when a function is defined.

### Basic Decorator
A decorator receives a function and returns a new function that can do extra work.

```python
def announce(function):
    def wrapper():
        print("Starting")
        function()
        print("Finished")
    return wrapper

@announce
def make_sound():
    print("Beep")

make_sound()
```

### Multiple Decorator Calls
After decoration, each call runs the added behavior too.

```python
make_sound()
make_sound()
```

### Arguments in the Decorated Function
The wrapper can accept and pass on the decorated function's arguments.

```python
def announce_call(function):
    def wrapper(name):
        print("Calling the function")
        return function(name)
    return wrapper

@announce_call
def greet_person(name):
    print("Hello,", name)

greet_person("Ari")
```

### Using `*args` and `**kwargs` in a Decorator
Use `*args` and `**kwargs` in the wrapper when it should accept different arguments.

```python
def trace(function):
    def wrapper(*args, **kwargs):
        print("Running", function.__name__)
        return function(*args, **kwargs)
    return wrapper

@trace
def add(first, second):
    return first + second

print(add(2, second=3))
```

### Decorator with Arguments
A decorator that takes its own setting needs an outer function to receive that setting.

```python
def repeat(times):
    def decorator(function):
        def wrapper(*args, **kwargs):
            for _ in range(times):
                function(*args, **kwargs)
        return wrapper
    return decorator

@repeat(2)
def say_hi():
    print("Hi")

say_hi()
```

### Multiple Decorators
You can place more than one decorator above a function. Python applies them from the closest decorator upward.

```python
def add_start(function):
    def wrapper():
        print("Start")
        function()
    return wrapper

def add_end(function):
    def wrapper():
        function()
        print("End")
    return wrapper

@add_start
@add_end
def task():
    print("Task")

task()
```

### Preserving Function Metadata
A wrapper can hide the original function's name and documentation. Use `functools.wraps` to preserve them.

```python
from functools import wraps

def simple_decorator(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        return function(*args, **kwargs)
    return wrapper

@simple_decorator
def welcome():
    """Print a welcome message."""
    print("Welcome")

print(welcome.__name__)
print(welcome.__doc__)
welcome()
```

## Python Generators
A generator is a kind of iterator that gives values one at a time. It can be useful when you want to process values as you need them.

### The `yield` Keyword
A function with `yield` produces a generator. Each time the generator is asked for a value, it continues until the next `yield`.

```python
def count_up_to_three():
    yield 1
    yield 2
    yield 3

for number in count_up_to_three():
    print(number)
```

### Generators Save Memory
A generator creates each value when it is needed instead of keeping every result in a list. This can save memory when working with many values.

```python
def squares(count):
    for number in range(count):
        yield number * number

for square_value in squares(4):
    print(square_value)
```

### Using `next()` with Generators
Call `next()` to get the next value from a generator. When there are no values left, Python raises `StopIteration`.

```python
numbers = count_up_to_three()
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

### Generator Expressions
A generator expression looks like a list comprehension with parentheses. It produces values one at a time.

```python
cubes = (number ** 3 for number in range(4))
for cube in cubes:
    print(cube)
```

### Fibonacci Sequence Generator
A generator can produce a sequence step by step. This one keeps creating Fibonacci numbers when requested.

```python
def fibonacci_generator():
    first, second = 0, 1
    while True:
        yield first
        first, second = second, first + second

sequence = fibonacci_generator()
for _ in range(7):
    print(next(sequence))
```

### Generator Methods
Generators have methods that let you send a value into a generator or stop it.

#### The `send()` Method
The `send()` method resumes a generator and sends a value into it. Start the generator with `next()` before sending a non-`None` value.

```python
def receive_value():
    value = yield "Ready"
    yield "Received: " + value

receiver = receive_value()
print(next(receiver))
print(receiver.send("hello"))
```

#### The `close()` Method
Call `close()` to stop a generator. It cannot produce more values after it is closed.

```python
def simple_generator():
    yield "First value"
    yield "Second value"

items = simple_generator()
print(next(items))
items.close()
```
