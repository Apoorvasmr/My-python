# Python Inheritance
Inheritance lets one class use properties and methods from another class. The class that gives these is the parent class; the class that receives them is the child class.

## Create a Parent Class
A parent class is a regular class. Its properties and methods can be used by child classes.

```python
class Animal:
    def speak(self):
        print("The animal makes a sound.")
```

## Create a Child Class
Put the parent class name in parentheses when defining a child class. The child can use methods from the parent.

```python
class Animal:
    def speak(self):
        print("The animal makes a sound.")

class Dog(Animal):
    pass

dog = Dog()
dog.speak()
```

## Add the `__init__()` Function
A child class can define its own `__init__()` method. This replaces the parent's initializer for child objects, so call `super().__init__()` if the parent also needs to set up its properties.

```python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

dog = Dog("Pip", "Collie")
print(dog.name)
print(dog.breed)
```

## Use the `super()` Function
`super()` gives the child access to methods from its parent class. It is often used to call the parent's `__init__()` method.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def describe(self):
        print(self.name, "is an animal.")

class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)

dog = Dog("Pip")
dog.describe()
```

## Add Properties
A child class can have its own properties in addition to the properties it gets from its parent.

```python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

dog = Dog("Pip", "Collie")
print(dog.name)
print(dog.breed)
```

## Add Methods
A child class can define its own methods. It can also use methods inherited from the parent.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name, "makes a sound.")

class Dog(Animal):
    def fetch(self):
        print(self.name, "fetches the ball.")

dog = Dog("Pip")
dog.speak()
dog.fetch()
```
