# Python Arrays
An array stores several values under one name. Python lists are often used for this purpose.

Python does not have a built-in array type like some languages. For typed arrays, Python also provides the `array` module, and NumPy is used for larger numerical arrays.

## What Is an Array?
An array is a collection of values that you can access by position. A list is a convenient way to store and work with a group of values in Python.

```python
cars = ["Toyota", "Honda", "Ford"]
print(cars)
```

### Using Python's `array` Module
Use the `array` module when you need an array that stores values of one type. Its type code indicates what kind of values it can hold.

```python
from array import array

scores = array("i", [10, 20, 30])
print(scores)
```

## Accessing Array Elements
Use an index in square brackets to get an item. Indexing starts at `0`, so the first item is at index `0`.

```python
cars = ["Toyota", "Honda", "Ford"]
print(cars[0])
print(cars[1])
```

You can change a list item by assigning a new value at its index.

```python
cars = ["Toyota", "Honda", "Ford"]
cars[0] = "Mazda"
print(cars)
```

## The Length of an Array
Use `len()` to find how many items are in a list. The length is one more than the highest index.

```python
cars = ["Toyota", "Honda", "Ford"]
print(len(cars))
```

## Looping Through Array Elements
Use a `for` loop to visit each item in a list.

```python
cars = ["Toyota", "Honda", "Ford"]
for car in cars:
    print(car)
```

## Adding Array Elements
Use `append()` to add an item to the end of a list.

```python
cars = ["Toyota", "Honda"]
cars.append("Ford")
print(cars)
```

## Removing Array Elements
Use `pop(index)` to remove an item by its position. If you leave out the index, `pop()` removes and returns the last item.

```python
cars = ["Toyota", "Honda", "Ford"]
removed_car = cars.pop(1)
print(removed_car)
print(cars)
```

Use `remove(value)` to remove the first item with that value. If the value is not in the list, Python raises an error.

```python
cars = ["Toyota", "Honda", "Toyota"]
cars.remove("Toyota")
print(cars)
```

## Array Methods
When using lists as arrays, you can use list methods to work with the items.

| Method | What it does |
| --- | --- |
| `append(item)` | Adds an item to the end. |
| `clear()` | Removes all items. |
| `copy()` | Makes a shallow copy. |
| `count(item)` | Counts matching items. |
| `extend(items)` | Adds items from another iterable. |
| `index(item)` | Finds the index of the first matching item. |
| `insert(index, item)` | Adds an item at a position. |
| `pop(index)` | Removes and returns an item. |
| `remove(item)` | Removes the first matching item. |
| `reverse()` | Reverses the item order. |
| `sort()` | Sorts the list in place. |

```python
numbers = [3, 1, 2, 1]
print(numbers.count(1))
numbers.sort()
print(numbers)
numbers.reverse()
print(numbers)
```
