# Python OOP
OOP means Object-Oriented Programming. It is a way to organize a program using classes and objects.

## What Is OOP?
In OOP, a class describes a kind of thing, and objects are individual things made from that class. Objects can hold data and have actions.

```python
class Pet:
    pass

first_pet = Pet()
second_pet = Pet()

print(type(first_pet))
print(first_pet is second_pet)
```

## Advantages of OOP
Classes can group related data and actions together. You can create many objects from one class and reuse the same structure instead of repeating code. This helps keep larger programs organized and supports the DRY idea: do not repeat the same code unnecessarily.

```python
class Pet:
    def speak(self):
        print("Hello!")

pets = [Pet(), Pet()]
for pet in pets:
    pet.speak()
```

## What Are Classes and Objects?
A class is a description used to create objects. An object is one instance of a class. Objects made from the same class share its structure, while each object is separate.

```python
class Book:
    category = "Book"

book_one = Book()
book_two = Book()

book_one.title = "The Blue Door"
book_two.title = "A Small Planet"

print(book_one.category, "-", book_one.title)
print(book_two.category, "-", book_two.title)
```
