## List in Python
A list is a collection of items stored in a single variable. Lists are ordered, changeable, and can contain duplicate values.

List syntax:
```python
my_list = ["apple", "banana", "cherry"]
```


````python
fruits = ["apple", "banana", "cherry"]
print(fruits)
print(len(fruits))
print(fruits[0])
````

## Creating a list
A list can be created using square brackets. The items inside can be strings, numbers, or even mixed data types.


````python
numbers = [10, 20, 30]
mixed = ["A", 1, True]
print(numbers)
print(mixed)
````

## Accessing list items
List items are accessed by index. Python indexing starts from 0.


````python
colors = ["red", "green", "blue"]
print(colors[0])
print(colors[1])
print(colors[-1])
````

## Slicing a list
Slicing returns a part of the list using the start and end index.


````python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
````

## Changing list items
We can change the value of any list item using its index.


````python
students = ["Aman", "Riya", "Karan"]
students[1] = "Meena"
print(students)
````

## List methods
Python lists have many built-in methods to add, remove, sort, search, and manipulate elements.


````python
# Example of list methods will be shown below.
print("List methods examples start here.")
````

## append()
Definition: Adds an element to the end of the list.


````python
students = ["Aman", "Riya"]
students.append("Karan")
print(students)
````

## clear()
Definition: Removes all elements from the list.


````python
numbers = [10, 20, 30]
numbers.clear()
print(numbers)
````

## copy()
Definition: Returns a copy of the list.


````python
numbers = [10, 20, 30]
copy_list = numbers.copy()
print(copy_list)
````

## count()
Definition: Returns the number of times a specified value appears in the list.


````python
numbers = [1, 2, 2, 3, 2]
print(numbers.count(2))
````

## extend()
Definition: Adds elements from another list to the end of the current list.


````python
list1 = ["a", "b"]
list2 = ["c", "d"]
list1.extend(list2)
print(list1)
````

## index()
Definition: Returns the index of the first matching value in the list.


````python
fruits = ["apple", "banana", "cherry"]
print(fruits.index("banana"))
````

## insert()
Definition: Inserts an element at a specified position.


````python
fruits = ["apple", "cherry"]
fruits.insert(1, "banana")
print(fruits)
````

## pop()
Definition: Removes and returns the item at the specified index.


````python
fruits = ["apple", "banana", "cherry"]
print(fruits.pop(1))
print(fruits)
````

## remove()
Definition: Removes the first matching element from the list.


````python
fruits = ["apple", "banana", "cherry"]
fruits.remove("banana")
print(fruits)
````

## reverse()
Definition: Reverses the order of the list.


````python
numbers = [1, 2, 3, 4]
numbers.reverse()
print(numbers)
````

## sort()
Definition: Sorts the list in ascending order by default.


````python
numbers = [5, 2, 8, 1]
numbers.sort()
print(numbers)
````
