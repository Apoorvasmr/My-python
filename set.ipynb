## Python Sets
A set is a collection that stores multiple items in one variable. Sets are unordered, changeable, and do not allow duplicate values. They are useful when you need to store unique items and check membership quickly.



```python
fruits = {"apple", "banana", "cherry"}
print(fruits)
```

## Sets are unordered
A set does not keep items in a fixed order. Because of that, you cannot access items by index like you can with a list or tuple.



```python
numbers = {5, 2, 7, 1}
print(numbers)
print(numbers)
```

## Duplicate values are ignored
If you add the same item more than once to a set, it only appears once. Python removes duplicates automatically.



```python
items = {"apple", "banana", "cherry", "apple", "banana"}
print(items)
print(len(items))
```

## Set length and type
Use `len()` to count how many unique items a set contains. Use `type()` to check that the object is a set.



```python
numbers = {1, 2, 3, 4}
print(len(numbers))
print(type(numbers))
```

## Accessing set items
You cannot access set items by index, but you can loop through the set or check whether a value is inside it with `in`.



```python
fruits = {"apple", "banana", "cherry"}
for x in fruits:
    print(x)

print("banana" in fruits)
print("mango" in fruits)
```

## Adding items to a set
Use `add()` to add one item, and `update()` to add multiple items from another iterable.



```python
fruits = {"apple", "banana"}
fruits.add("cherry")
print(fruits)

more_fruits = ["orange", "mango"]
fruits.update(more_fruits)
print(fruits)
```

## Removing items from a set
You can remove values with `remove()`, `discard()`, or `pop()`. `clear()` removes everything, and `del` deletes the whole set.



```python
fruits = {"apple", "banana", "cherry", "date"}
fruits.remove("banana")
print(fruits)

fruits.discard("mango")
print(fruits)

x = fruits.pop()
print(x)
print(fruits)

fruits.clear()
print(fruits)
```

## Set operations: union, intersection, and difference
Set operations let you combine or compare sets. These are powerful when working with groups of unique values.



```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(set_a | set_b)
print(set_a & set_b)
print(set_a - set_b)
print(set_a ^ set_b)
```

## Set methods summary
Python provides many built-in set methods such as `union()`, `intersection()`, `difference()`, `copy()`, `issubset()`, and `issuperset()`.



```python
A = {1, 2, 3}
B = {2, 3, 4}
C = {1, 2}

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(C.issubset(A))
print(A.issuperset(C))
print(A.copy())
```

## Practical set example
A set is great for storing unique values, checking membership, and performing mathematical set operations.



```python
students = {"Alice", "Bob", "Charlie", "Alice"}
print(students)

all_students = {"Alice", "Bob", "David"}
new_students = {"Charlie", "Eva"}

print(all_students | new_students)
print(all_students & students)
print(students - all_students)
print("Alice" in students)
```
