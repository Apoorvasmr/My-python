# Python `self` Parameter
In a class method, `self` refers to the current object. It lets the method work with that object's data and other methods.

## The `self` Parameter
Write `self` as the first parameter of an instance method. When you call the method through an object, Python passes that object as `self` automatically.

```python
class Greeter:
    def greet(self):
        print("Hello from this object!")

greeter = Greeter()
greeter.greet()
```

## Why Use `self`?
Use `self` to connect a method to the object that called it. This lets one class method behave differently for different objects.

```python
class Greeter:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello,", self.name)

Greeter("Ari").greet()
Greeter("Bea").greet()
```

## `self` Does Not Have to Be Named `self`
Python allows another name for the first method parameter, but `self` is the standard name. Using it makes code easier for other Python programmers to understand.

```python
class Greeter:
    def greet(current_object):
        print("Hello from this object!")

Greeter().greet()
```

## Accessing Properties with `self`
Use `self.property_name` to read or change a value stored on the current object.

```python
class Counter:
    def __init__(self, start):
        self.value = start

    def show(self):
        print(self.value)

counter = Counter(3)
counter.show()
```

## Calling Methods with `self`
A method can call another method on the same object by writing `self.method_name()`.

```python
class Greeter:
    def greet(self):
        print(self.make_message())

    def make_message(self):
        return "Welcome!"

Greeter().greet()
```
