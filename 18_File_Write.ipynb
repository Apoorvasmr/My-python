# Writing to Files
Python can write text to files. This chapter explains how to add text, replace file contents, and create a new file. Examples use temporary folders so they do not change your existing files.

## Write to an Existing File
Use `open(path, "a")` to add text to the end of a file. Use `open(path, "w")` to replace its contents. Both modes also create a file if it does not exist. Use `write()` to add text; it does not add a newline unless you include `\n`.

### Add to the End of a File
Mode `"a"` keeps the current contents and adds new text at the end.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as folder:
    path = Path(folder) / "notes.txt"
    path.write_text("First line\n", encoding="utf-8")

    with open(path, "a", encoding="utf-8") as file:
        file.write("Second line\n")

    print(path.read_text(encoding="utf-8"))
```

### Overwrite Existing Content
Mode `"w"` removes the old contents before writing. Be careful to use it only when replacing the contents is intended.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "draft.txt"
    path.write_text("Old contents", encoding="utf-8")

    with open(path, "w", encoding="utf-8") as file:
        file.write("New contents")

    print(path.read_text(encoding="utf-8"))
```

## Create a New File
Use mode `"x"` to create a new file only. It raises `FileExistsError` if the path already exists. Mode `"w"` also creates a file when needed, but replaces it if it already exists.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "new_file.txt"

    with open(path, "x", encoding="utf-8") as file:
        file.write("This file was created.")

    print(path.read_text(encoding="utf-8"))
```

You can also create a file with mode `"w"`. If the file exists, its old contents are replaced.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "another_file.txt"

    with open(path, "w", encoding="utf-8") as file:
        file.write("Created with write mode.")

    print(path.read_text(encoding="utf-8"))
```
