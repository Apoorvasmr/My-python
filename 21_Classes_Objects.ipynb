# Python Classes and Objects
A class is a description used to create objects. An object is one item made from a class.

## Python Classes and Objects
A class can describe what its objects have and what they can do. Each object created from the class is a separate instance.

## Create a Class
Use the `class` keyword followed by a class name. Class names usually start with a capital letter.

```python
class Book:
    category = "Story"
```

## Create an Object
Call the class name with parentheses to create an object. Use dot notation to access its class properties.

```python
class Book:
    category = "Story"

book = Book()
print(book.category)
```

## Delete Objects
Use `del` to remove a name that refers to an object. This removes that reference; Python cleans up the object when it is no longer in use.

```python
class Book:
    pass

book = Book()
del book
print("The book variable was deleted.")
```

## Multiple Objects
You can create many objects from one class. Each object can have its own values.

```python
class Book:
    pass

first_book = Book()
first_book.title = "The Blue Door"

second_book = Book()
second_book.title = "A Small Planet"

print(first_book.title)
print(second_book.title)
```

## The `pass` Statement
A class body cannot be empty. Use `pass` as a placeholder when you want to define the class but add its contents later.

```python
class EmptyClass:
    pass

example = EmptyClass()
print(type(example).__name__)
```
