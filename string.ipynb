# Python Strings

## String
A string is a sequence of characters used to store text. In Python, a string is represented by the `str` data type. Strings are used for names, messages, file paths, and user input.

### Definition
A string is a text value enclosed in quotes. Examples include `'hello'`, "Python", and """multi-line text""".

### Syntax
Python strings can be written with:
- single quotes: `'hello'`
- double quotes: "hello"
- triple quotes: """hello"""

Triple quotes are useful for multi-line text.


```python
# String syntax example
name = 'Python'
message = "Hello World"
paragraph = """This is a multi-line string.
It continues on the next line."""

print(name)
print(message)
print(paragraph)
```

## Indexing and Slicing
Strings are ordered, and each character has a position called an index. Indexing starts from 0, so the first character is at position 0.

- `text[0]` gets the first character
- `text[-1]` gets the last character
- `text[start:end]` gets a slice of the string

Slicing helps us take part of a string without changing the original value.


```python
word = "Python"
print(word[0])
print(word[-1])
print(word[0:3])
print(word[:3])
print(word[3:])
```

## Concatenation and Formatting
String concatenation means joining strings together with `+`.

Formatting means inserting values into a string using `.format()` or f-strings.

This is useful when creating messages or reports.


```python
first = "Hello"
last = "World"
name = "Aisha"

print(first + " " + last)
print("Hello, {}!".format(name))
print(f"Welcome {name} to Python class.")
```

## Escape Sequences
Escape sequences begin with a backslash (`\`) and are used to add special characters to a string.

| Escape | Meaning |
| --- | --- |
| `\'` | Single quote |
| `\\` | Backslash |
| `\n` | New line |
| `\r` | Carriage return |
| `\t` | Tab |
| `\b` | Backspace |
| `\f` | Form feed |
| `\ooo` | Octal value |
| `\xhh` | Hex value |

These sequences are useful for formatting text, file paths, and messages.


```python
print("Hello\nWorld")
print("Tab\tExample")
print("Single quote: \'A\'")
print("Backslash: \\")
print("Carriage return: \rX")
print("Octal: \101")
print("Hex: \x41")
```
