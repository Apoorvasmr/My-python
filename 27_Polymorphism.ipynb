# Python Polymorphism
Polymorphism means that the same function, method, or operation can work with different types of objects.

## Function Polymorphism
A built-in function can work with different types. For example, `len()` returns the number of characters in a string and the number of items in a collection.

```python
print(len("hello"))
print(len(["red", "blue", "green"]))
```

## String
The `+` operator joins strings together. The same operator adds numbers, so what it does depends on the values.

```python
first = "Good"
second = " morning"
print(first + second)
print(4 + 3)
```

## Tuple
The `+` operator joins tuples to make a new tuple.

```python
first_pair = (1, 2)
second_pair = (3, 4)
print(first_pair + second_pair)
```

## Dictionary
The `+` operator does not join dictionaries. Use the dictionary update operation to copy entries from one dictionary into another.

```python
person = {"name": "Ari"}
details = {"city": "Delhi"}

person.update(details)
print(person)
```

## Class Polymorphism
Different classes can have methods with the same name. The method that runs depends on the object.

```python
class Bird:
    def speak(self):
        print("Tweet")

class Dog:
    def speak(self):
        print("Woof")

for animal in (Bird(), Dog()):
    animal.speak()
```

## Inheritance Class Polymorphism
A child class can replace a method from its parent with its own version. This is called overriding. The same method call can then do different things for different child objects.

```python
class Animal:
    def speak(self):
        print("Some sound")

class Bird(Animal):
    def speak(self):
        print("Tweet")

class Dog(Animal):
    def speak(self):
        print("Woof")

for animal in (Bird(), Dog()):
    animal.speak()
```
