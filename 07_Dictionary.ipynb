## Python Dictionaries
A dictionary is a collection of data values stored in key:value pairs. Each key is unique, and it is used to access its corresponding value. Dictionaries are ordered in Python 3.7+, changeable, and they do not allow duplicate keys.

```python
car = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(car)
```

## Dictionary items
Dictionary items are stored as key:value pairs. You can access a value by writing the dictionary name followed by the key in square brackets.

```python
car = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(car["brand"])
print(car["model"])
print(car["year"])
```

## Access with get()
The `get()` method returns the value of a specified key. It is a safe way to access values without raising a `KeyError` if the key does not exist.

```python
car = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(car.get("model"))
print(car.get("color", "Not found"))
print(car.get("year"))
```

## Change values
You can change the value of a specific key by assigning a new value to it. The dictionary is updated in place.

```python
car = {"brand": "Ford", "model": "Mustang", "year": 1964}
car["year"] = 2020
car["model"] = "Focus"
print(car)
```

## Update a dictionary
The `update()` method can add or change items in one call. It accepts another dictionary or an iterable of key:value pairs.

```python
car = {"brand": "Ford", "model": "Mustang"}
car.update({"year": 2020, "color": "red"})
car.update({"model": "Focus"})
print(car)
```

## Add items
New items are added by writing a new key and assigning a value to it. This makes the dictionary grow dynamically.

```python
car = {"brand": "Ford", "model": "Mustang"}
car["color"] = "white"
car["brand"] = "Chevrolet"
print(car)
```

## Remove items
Use `pop()`, `popitem()`, `clear()`, or `del` to remove dictionary entries. `pop()` removes a specific key, while `clear()` empties the dictionary.

```python
car = {"brand": "Ford", "model": "Mustang", "year": 1964}

car.pop("model")
print(car)

car.popitem()
print(car)

car.clear()
print(car)

del car
print(car)  
```

## Loop through a dictionary
A `for` loop can iterate over the keys, values, or key:value pairs in a dictionary. The `keys()`, `values()`, and `items()` methods help with this.

```python
student = {"name": "Ali", "age": 20, "grade": "A"}
for key in student:
    print(key)

for value in student.values():
    print(value)

for key, value in student.items():
    print(key, value)
```

## Copy a dictionary
A dictionary copy is not the same as a reference. If you assign one dictionary to another, both names point to the same object. Use `copy()` or `dict()` to make a separate copy.

```python
car = {"brand": "Ford", "model": "Mustang"}
new_car = car.copy()
new_car["year"] = 2024
print(car)
print(new_car)
```

## Nested dictionaries
A dictionary can contain other dictionaries. This is useful when you want to group related data together.

```python
family = {
    "child1": {"name": "Emil", "year": 2004},
    "child2": {"name": "Tobias", "year": 2007},
    "child3": {"name": "Linus", "year": 2011}
}
print(family["child2"]["name"])
print(family["child3"]["year"])
```

## Dictionary methods summary
Python dictionaries include several useful built-in methods such as `keys()`, `values()`, `items()`, `get()`, `update()`, `copy()`, and `clear()`.

```python
student = {"name": "Ali", "age": 20, "course": "Python"}
print(student.keys())
print(student.values())
print(student.items())
print(student.get("name"))
```
