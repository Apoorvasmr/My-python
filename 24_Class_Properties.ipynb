# Python Class Properties
Properties are values stored on a class or on an object. They describe information about that class or object.

## Class Properties
A class property is written in the class body, outside its methods. Objects made from the class can use it.

```python
class Car:
    wheels = 4

car = Car()
print(car.wheels)
```

## Access Properties
Use dot notation to read a property from a class or an object.

```python
class Car:
    wheels = 4

car = Car()
print(Car.wheels)
print(car.wheels)
```

## Modify Properties
Assign a new value with dot notation to change a property on an object.

```python
class Car:
    color = "blue"

car = Car()
car.color = "red"
print(car.color)
```

## Delete Properties
Use `del` to remove a property from an object. If a class has a property with the same name, the object can use the class value again.

```python
class Car:
    color = "blue"

car = Car()
car.color = "red"
del car.color
print(car.color)
```

## Class Properties vs Object Properties
A class property belongs to the class and can be shared by its objects. An object property belongs to one object. An object property with the same name is used instead of the class property for that object.

```python
class Car:
    wheels = 4  # class property

first_car = Car()
second_car = Car()
first_car.wheels = 6  # property on this object

print(first_car.wheels)
print(second_car.wheels)
print(Car.wheels)
```

## Modifying Class Properties
Assign to the class name to change the class property. Objects that do not have their own property with that name will see the new value.

```python
class Car:
    wheels = 4

first_car = Car()
second_car = Car()
Car.wheels = 3

print(first_car.wheels)
print(second_car.wheels)
```

## Add New Properties
You can add a property to an object by assigning to a new name. That property belongs only to that object.

```python
class Car:
    pass

car = Car()
car.color = "green"
print(car.color)
```
