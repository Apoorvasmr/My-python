# Python Class Methods
A method is a function defined inside a class. It describes an action that an object can perform.

## Class Methods
Define a method inside the class using `def`. An instance method uses `self` as its first parameter.

```python
class Greeter:
    def greet(self):
        print("Hello!")

greeter = Greeter()
greeter.greet()
```

## Methods with Parameters
A method can take extra parameters after `self`. Give those values when you call the method.

```python
class Greeter:
    def greet(self, name):
        print("Hello,", name)

greeter = Greeter()
greeter.greet("Ari")
```

## Methods Accessing Properties
Use `self.property_name` inside a method to read a property from the object.

```python
class Pet:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print("My name is", self.name)

Pet("Milo").introduce()
```

## Methods Modifying Properties
A method can assign a new value to a property using `self.property_name`.

```python
class Counter:
    def __init__(self):
        self.value = 0

    def add_one(self):
        self.value += 1

counter = Counter()
counter.add_one()
print(counter.value)
```

## The `__str__()` Method
Define `__str__()` to choose the text shown when you print an object. It should return a string.

```python
class Pet:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return "Pet: " + self.name

print(Pet("Milo"))
```

## Multiple Methods
A class can have several methods. One method can call another method on the same object with `self.method_name()`.

```python
class Pet:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(self.greeting())

    def greeting(self):
        return "Hello, I am " + self.name

Pet("Luna").introduce()
```

## Delete Methods
Python does not provide a simple way to remove one method from a class instance. You can delete the class attribute, but this changes the class for every object and is rarely needed. This example uses a temporary class so the change does not affect other code.

```python
class TemporaryGreeter:
    def greet(self):
        print("Hello!")

del TemporaryGreeter.greet
print("The greet method was removed from TemporaryGreeter.")
```
