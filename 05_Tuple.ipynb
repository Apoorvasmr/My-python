## Python Tuples
A tuple is an ordered collection that stores multiple items in one variable. Unlike a list, a tuple is immutable: its items cannot be changed, added, or removed after creation. Tuples allow duplicate values and use indexes starting at 0.

Tuple syntax:
```python
my_tuple = ("apple", "banana", "cherry")
```


````python
fruits = ("apple", "banana", "cherry")
print(fruits)

colors = "red", "green", "blue"
print(colors)
````

## Tuples are ordered
A tuple is ordered, so its items have a fixed position. The order stays the same unless you create a different tuple.


````python
colors = ("red", "green", "blue")
print(colors[0])
print(colors[1])
print(colors[2])
````

## Tuples are unchangeable
Tuples are immutable, so you cannot assign to an item, add an item, or remove an item after the tuple is created. You can make a new tuple instead.


````python
colors = ("red", "green", "blue")
new_colors = colors + ("yellow",)
print(colors)
print(new_colors)
````

## Tuples allow duplicate values
A tuple can contain the same value more than once. Each occurrence has its own index and contributes to the tuple's length.


````python
fruits = ("apple", "banana", "cherry", "apple", "cherry")
print(fruits)
print(len(fruits))
````

## Creating a tuple
Write tuple items inside round brackets and separate them with commas. Python also allows comma-separated values without parentheses.


````python
fruits = ("apple", "banana", "cherry")
colors = "red", "green", "blue"
print(fruits)
print(colors)
````

## Tuple length
Use the built-in `len()` function to find how many items are in a tuple. Repeated values are counted separately.


````python
fruits = ("apple", "banana", "apple")
print(len(fruits))
````

## Create a one-item tuple
A one-item tuple must include a trailing comma after its item. Without the comma, parentheses only group the value and do not create a tuple.


````python
single_item_tuple = ("apple",)
not_a_tuple = ("apple")
print(type(single_item_tuple))
print(type(not_a_tuple))
````

## Create an empty tuple
Use empty round brackets `()` to create a tuple with no items. Its length is zero.


````python
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
print(len(empty_tuple))
````

## Tuple item data types
Tuple items can be values of any data type, and a tuple can mix different types. The tuple's order and immutability do not depend on item types.


````python
strings = ("apple", "banana")
numbers = (1, 2, 3)
flags = (True, False)
mixed = ("age", 20, True)
print(strings)
print(numbers)
print(flags)
print(mixed)
````

## Tuple type
The built-in `type()` function identifies an object's data type. A tuple has the type `tuple`.


````python
fruits = ("apple", "banana", "cherry")
print(type(fruits))
````

## The tuple() constructor
The `tuple()` constructor creates a tuple from an iterable, such as a list or a string. Passing a string creates a tuple containing one item for each character.


````python
fruits = tuple(["apple", "banana", "cherry"])
letters = tuple("cat")
print(fruits)
print(letters)
````

## Accessing tuple items
Access an item by placing its index in square brackets. Index `0` is the first item; negative indexes count backward from the end, with `-1` for the last item.


````python
colors = ("red", "green", "blue")
print(colors[0])
print(colors[1])
print(colors[-1])
print(colors[-2])
````

## Range of tuple indexes
A slice uses `tuple[start:stop]` to return a new tuple. The start index is included and the stop index is excluded. Either bound can be omitted, and negative indexes can be used.


````python
fruits = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(fruits[2:5])
print(fruits[:4])
print(fruits[2:])
print(fruits[-4:-1])
````

## Check whether an item exists
Use the `in` operator to test whether a value is present in a tuple. It returns `True` or `False`.


````python
fruits = ("apple", "banana", "cherry")
print("banana" in fruits)
print("mango" in fruits)
````

## Changing tuple values
A tuple cannot be edited in place. To change a value, convert the tuple to a list, update the list, and convert it back to a tuple.


````python
items = ("apple", "banana", "cherry")
items_as_list = list(items)
items_as_list[1] = "kiwi"
items = tuple(items_as_list)
print(items)
````

## Adding items to a tuple
Tuples have no `append()` method. To add items, convert the tuple to a list, add the values, and convert the list back to a tuple.


````python
items = ("apple", "banana", "cherry")
items_as_list = list(items)
items_as_list.append("orange")
items = tuple(items_as_list)
print(items)
````

## Adding tuples with the + operator
You can combine tuples with `+`. Since tuples are immutable, this creates a new tuple rather than changing the original tuple in place.


````python
fruits = ("apple", "banana", "cherry")
new_item = ("orange",)
fruits += new_item
print(fruits)
````

## Removing items from a tuple
Tuples cannot remove items in place. Convert a tuple to a list, remove the item from that list, and convert the result back to a tuple.


````python
items = ("apple", "banana", "cherry")
items_as_list = list(items)
items_as_list.remove("banana")
items = tuple(items_as_list)
print(items)
````

## Delete a tuple variable
The `del` keyword removes the tuple variable itself. It does not remove one item; after deletion, the variable name is no longer defined.


````python
items = ("apple", "banana", "cherry")
del items
print("items" in locals())
````

## Packing a tuple
Packing means placing multiple values into a single tuple. Python creates the tuple from the comma-separated values.


````python
fruits = "apple", "banana", "cherry"
print(fruits)
````

## Unpacking a tuple
Unpacking assigns each tuple item to a separate variable. The number of variables must match the number of items unless starred unpacking is used.


````python
fruits = ("apple", "banana", "cherry")
(first, second, third) = fruits
print(first)
print(second)
print(third)
````

## Unpacking with an asterisk
Use `*` before one variable to collect multiple remaining items into a list. The starred variable can be placed between other variables.


````python
fruits = ("apple", "banana", "cherry", "orange", "kiwi")
(first, second, *remaining) = fruits
print(first)
print(second)
print(remaining)

(first, *middle, last) = fruits
print(first)
print(middle)
print(last)
````

## Loop through a tuple
A `for` loop visits each tuple item in order. Use this form when you need the values but not their index numbers.


````python
fruits = ("apple", "banana", "cherry")
for fruit in fruits:
    print(fruit)
````

## Loop through tuple indexes
Use `range(len(tuple_name))` to generate each valid index and access the corresponding item.


````python
fruits = ("apple", "banana", "cherry")
for index in range(len(fruits)):
    print(index, fruits[index])
````

## Loop through a tuple with while
A `while` loop can access tuple items by index. Start at index `0` and increase the index each time until it reaches the tuple's length.


````python
fruits = ("apple", "banana", "cherry")
index = 0
while index < len(fruits):
    print(fruits[index])
    index += 1
````

## Joining two tuples
Use the `+` operator to join tuples. It returns a new tuple and leaves the original tuples unchanged.


````python
letters = ("a", "b", "c")
numbers = (1, 2, 3)
combined = letters + numbers
print(combined)
print(letters)
````

## Multiplying a tuple
The `*` operator repeats the tuple's items the specified number of times and returns a new tuple.


````python
fruits = ("apple", "banana", "cherry")
repeated_fruits = fruits * 2
print(repeated_fruits)
````

## count()
The `count()` method returns the number of times a specified value appears in a tuple.


````python
numbers = (1, 2, 2, 3, 2)
print(numbers.count(2))
````

## index()
The `index()` method returns the position of the first matching value in a tuple. It raises `ValueError` if the value is not present.


````python
fruits = ("apple", "banana", "cherry", "banana")
print(fruits.index("banana"))
````
