# Python `__init__()` Method
The `__init__()` method runs automatically when you create a new object. It is often used to give the object its starting values.

## The `__init__()` Method
Define `__init__()` inside a class. Its first parameter is `self`, followed by any values the new object needs.

```python
class Pet:
    def __init__(self, name):
        self.name = name

pet = Pet("Milo")
print(pet.name)
```

## Why Use `__init__()`?
It lets each new object start with the values it needs. You do not have to create the object first and set those values in separate steps.

```python
class Pet:
    def __init__(self, name):
        self.name = name

first_pet = Pet("Milo")
second_pet = Pet("Luna")

print(first_pet.name)
print(second_pet.name)
```

## Default Values in `__init__()`
You can give a parameter a default value. Python uses it when the object is created without a value for that parameter.

```python
class Pet:
    def __init__(self, name="Unknown"):
        self.name = name

pet_one = Pet()
pet_two = Pet("Coco")

print(pet_one.name)
print(pet_two.name)
```

## Multiple Parameters
An `__init__()` method can take several parameters. Use them to set several starting values on the object.

```python
class Pet:
    def __init__(self, name, animal, age):
        self.name = name
        self.animal = animal
        self.age = age

pet = Pet("Milo", "cat", 3)
print(pet.name)
print(pet.animal)
print(pet.age)
```
