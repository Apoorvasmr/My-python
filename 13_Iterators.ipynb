# Python Iterators
An iterator gives items from a collection one at a time. Python uses iterators behind the scenes in `for` loops.

## Iterator vs Iterable
An iterable is something you can get an iterator from, such as a list, tuple, or string. An iterator is the object that provides the next item when asked.

```python
colors = ["red", "green", "blue"]  # iterable
color_iterator = iter(colors)       # iterator

print(next(color_iterator))
print(next(color_iterator))
```

Strings and tuples are also iterable. Calling `iter()` gives an iterator for each one.

```python
word_iterator = iter("cat")
tuple_iterator = iter((10, 20))

print(next(word_iterator))
print(next(tuple_iterator))
```

## Looping Through an Iterator
A `for` loop gets items from an iterator one by one and stops when there are no more items.

```python
animals = ["fox", "owl", "bear"]
for animal in iter(animals):
    print(animal)
```

You can also ask for each item with `next()`. Each call moves the iterator forward.

```python
numbers = iter([4, 5, 6])
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

## Create an Iterator
A custom iterator is a class that defines `__iter__()` and `__next__()`. The `__iter__()` method returns the iterator, and `__next__()` returns the next item.

```python
class CountToThree:
    def __iter__(self):
        self.number = 1
        return self

    def __next__(self):
        if self.number <= 3:
            current = self.number
            self.number += 1
            return current
        raise StopIteration

for number in CountToThree():
    print(number)
```

Calling `iter()` starts this example at `1`. Each call to `next()` gives the next number, and the iterator remembers its place.

```python
counter = iter(CountToThree())
print(next(counter))
print(next(counter))
```

## `StopIteration`
When an iterator has no more items, `next()` raises `StopIteration`. A `for` loop handles this automatically. A custom iterator raises it to tell Python that it is finished.

```python
counter = iter(CountToThree())
print(next(counter))
print(next(counter))
print(next(counter))

try:
    next(counter)
except StopIteration:
    print("The iterator is finished.")
```
