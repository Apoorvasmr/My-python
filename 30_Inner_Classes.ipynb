# Python Inner Classes
An inner class is a class defined inside another class. It can be useful for keeping closely related classes together.

## Python Inner Classes
Define the inner class inside the body of the outer class. The inner class is a separate class; it is not automatically connected to a particular outer object.

```python
class School:
    class Student:
        def introduce(self):
            print("I am a student.")

student = School.Student()
student.introduce()
```

## Accessing an Inner Class from the Outside
Use the outer class name, a dot, and the inner class name to create an inner-class object from outside.

```python
class School:
    class Student:
        def __init__(self, name):
            self.name = name

student = School.Student("Ari")
print(student.name)
```

## Accessing the Outer Class from the Inner Class
An inner-class object does not get the outer object automatically. Pass the outer object to the inner class if it needs to use outer-object properties.

```python
class School:
    def __init__(self, name):
        self.name = name
        self.student = self.Student("Ari", self)

    class Student:
        def __init__(self, name, school):
            self.name = name
            self.school = school

        def introduce(self):
            print(self.name, "studies at", self.school.name)

school = School("North School")
school.student.introduce()
```

## Practical Example
An inner class can describe a part that belongs to a larger object. In this example, a computer has a screen.

```python
class Computer:
    def __init__(self, screen_size):
        self.screen = self.Screen(screen_size)

    class Screen:
        def __init__(self, size):
            self.size = size

computer = Computer("14 inches")
print(computer.screen.size)
```

## Multiple Inner Classes
An outer class can contain more than one inner class when each represents a different related part.

```python
class Computer:
    class Screen:
        def show(self):
            print("Screen is on.")

    class Keyboard:
        def type_text(self):
            print("Typing...")

computer_screen = Computer.Screen()
computer_keyboard = Computer.Keyboard()
computer_screen.show()
computer_keyboard.type_text()
```
