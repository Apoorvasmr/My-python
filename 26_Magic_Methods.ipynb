# Python Magic Methods
Magic methods are special methods Python calls automatically in certain situations. Their names begin and end with two underscores, so they are also called dunder methods.

## What Are Magic Methods?
You usually do not call a magic method directly. Python calls it when you perform an action, such as creating or printing an object. For example, `__init__()` runs when an object is created, and `__str__()` runs when an object is turned into readable text.

```python
class Book:
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return "Book: " + self.title

book = Book("The Green Hill")  # Python calls __init__()
print(book)                    # Python calls __str__()
```

## List of Magic Methods
These methods let a class define how its objects behave in common Python operations.

| Method | Called when | What it controls |
| --- | --- | --- |
| `__init__()` | An object is created, like `Book(...)` | Sets up the new object. |
| `__str__()` | `print(obj)` or `str(obj)` is used | Readable text for the object. |
| `__repr__()` | `repr(obj)` is used | A representation useful for developers. |
| `__eq__()` | `obj1 == obj2` is used | How two objects are compared for equality. |
| `__add__()` | `obj1 + obj2` is used | What addition means for the objects. |
| `__len__()` | `len(obj)` is used | The object's length. |
| `__lt__()` | `obj1 < obj2` is used | How objects are ordered in a comparison. |
| `__contains__()` | `item in obj` is used | How Python checks whether an item is present. |
| `__call__()` | `obj(...)` is used | Lets the object be called like a function. |

## The `__str__()` Method
Python calls `__str__()` when you use `print()` or `str()` on an object. It must return a string for people to read.

### `__str__()` Must Return a String
Return text from `__str__()`. Returning another type causes Python to raise a `TypeError` when the object is printed.

```python
class Book:
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return "Book: " + self.title

print(Book("River Path"))
```

## The `__repr__()` Method
`repr(obj)` calls `__repr__()`. Its result is usually a clear representation for debugging. When practical, make it unambiguous.

```python
class Book:
    def __init__(self, title):
        self.title = title

    def __repr__(self):
        return f"Book({self.title!r})"

book = Book("River Path")
print(repr(book))
```

### `__str__()` vs `__repr__()`
`__str__()` is for friendly display. `__repr__()` is for a more detailed, developer-facing representation. If a class has no `__str__()`, Python can use `__repr__()` as a fallback.

```python
class Book:
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return self.title

    def __repr__(self):
        return f"Book({self.title!r})"

book = Book("River Path")
print(str(book))
print(repr(book))
```

## The `__eq__()` Method
Python calls `__eq__()` for `==`. Define it to say when two objects should count as equal.

### Change How to Compare
Compare the properties that matter for your class. This example treats books with the same title as equal.

```python
class Book:
    def __init__(self, title):
        self.title = title

    def __eq__(self, other):
        if not isinstance(other, Book):
            return NotImplemented
        return self.title == other.title

print(Book("River Path") == Book("River Path"))
```

## The `__add__()` Method
Python calls `__add__()` for the `+` operator. Choose what adding two objects means for your class.

```python
class Score:
    def __init__(self, points):
        self.points = points

    def __add__(self, other):
        if not isinstance(other, Score):
            return NotImplemented
        return Score(self.points + other.points)

total = Score(4) + Score(6)
print(total.points)
```

## The `__len__()` Method
Python calls `__len__()` when you use `len(obj)`. Return a non-negative integer that describes the object's length.

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

playlist = Playlist(["Song A", "Song B", "Song C"])
print(len(playlist))
```

## The `__lt__()` Method
Python calls `__lt__()` for the less-than operator (`<`). Define which property should be used to compare two objects.

### Compare Without `__lt__()`
Without a comparison method, Python generally cannot tell which of two custom objects is smaller. Use `sorted(..., key=...)` to sort by a chosen property instead.

```python
class Player:
    def __init__(self, score):
        self.score = score

players = [Player(8), Player(3), Player(5)]
ordered = sorted(players, key=lambda player: player.score)
print([player.score for player in ordered])
```

### Sorting with `__lt__()`
If `__lt__()` compares the objects, `sorted()` can use that rule to order them.

```python
class Player:
    def __init__(self, score):
        self.score = score

    def __lt__(self, other):
        if not isinstance(other, Player):
            return NotImplemented
        return self.score < other.score

players = sorted([Player(8), Player(3), Player(5)])
print([player.score for player in players])
```

### Sorting Without `__lt__()`
You do not need to add `__lt__()` just to sort objects. Give `sorted()` a `key` function that returns the value to sort by.

```python
class Player:
    def __init__(self, name):
        self.name = name

players = [Player("Zoe"), Player("Ari"), Player("Mina")]
ordered = sorted(players, key=lambda player: player.name)
print([player.name for player in ordered])
```

## The `__contains__()` Method
Python calls `__contains__()` for the `in` operator. Return `True` if the item is in the object and `False` otherwise.

```python
class Shelf:
    def __init__(self, books):
        self.books = books

    def __contains__(self, title):
        return title in self.books

shelf = Shelf(["River Path", "Cloud Atlas"])
print("River Path" in shelf)
print("Green Hill" in shelf)
```

### Without `__contains__()`
If the class has no `__contains__()`, Python can check items by iterating over the object when it is iterable. Defining `__contains__()` lets you choose a direct membership check.

```python
class Shelf:
    def __init__(self, books):
        self.books = books

    def __iter__(self):
        return iter(self.books)

shelf = Shelf(["River Path", "Cloud Atlas"])
print("Cloud Atlas" in shelf)
```

## The `__call__()` Method
Define `__call__()` to make an object callable like a function. Python runs this method when you write the object name followed by parentheses.

```python
class Multiplier:
    def __init__(self, amount):
        self.amount = amount

    def __call__(self, number):
        return self.amount * number

double = Multiplier(2)
print(double(5))
```
