# Python `range()`
A `range` represents a sequence of integers. It is often used to repeat code in a `for` loop.

The stop value is not included in the sequence. A range does not make a list of all its numbers at once.

## Creating Ranges
Use `range(stop)`, `range(start, stop)`, or `range(start, stop, step)`. The start value is included, but the stop value is not. The step tells how much to add each time.

### Call `range()` with One Argument
With one argument, the range starts at `0` and stops before the given number.

```python
numbers = range(5)
print(list(numbers))
```

### Call `range()` with Two Arguments
With two arguments, the first is the start and the second is the stop. The default step is `1`.

```python
numbers = range(3, 8)
print(list(numbers))
```

### Call `range()` with Three Arguments
With three arguments, the last one is the step. The step cannot be `0`. A negative step counts downward.

```python
by_twos = range(2, 11, 2)
backwards = range(5, 0, -1)

print(list(by_twos))
print(list(backwards))
```

## Using Ranges
A range is useful when you need to repeat something a set number of times or work with a sequence of numbers.

```python
for number in range(1, 4):
    print("Turn", number)
```

## Using a List to Display Ranges
Use `list()` when you want to see all the numbers in a range at once. This creates a list containing those numbers.

```python
numbers = list(range(4, 9))
print(numbers)
```

## Slicing Ranges
A range can be sliced like a list. A slice gives back another range and uses the same start-inclusive, stop-exclusive rule.

```python
numbers = range(10)
part = numbers[2:8:2]
print(list(part))
```

## Membership Testing
Use `in` to check whether a number is in a range. Use `not in` to check that it is not.

```python
even_numbers = range(2, 12, 2)
print(6 in even_numbers)
print(7 not in even_numbers)
```

## Length
Use `len()` to find how many numbers a range contains. An empty range has length `0`.

```python
numbers = range(2, 12, 2)
empty_range = range(5, 2)

print(len(numbers))
print(len(empty_range))
```
