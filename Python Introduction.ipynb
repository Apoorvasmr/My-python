# PYTHON

## Python:
Python is a high-level, interpreted, interactive, object - oriented programming language with automatic memory management and garbage collector. It is dynamically typed and loosely coupled.

## High level:
Python is an interpreted high level language. Internally python first compiles the high level code (.py) into bytecode (.pyc) then the interpreter executes that bytecode to generate the output.

## Interpreted:
An interpreter executes the code line by line, and if it encounters any error it stops the execution at that point.
