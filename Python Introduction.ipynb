# PYTHON

## Python:
Python is a high-level, interpreted, interactive, object - oriented programming language with automatic memory management and garbage collector. It is dynamically typed and loosely coupled.

- ### High level:
    - Python is an interpreted high level language. Internally python first compiles the high level code (.py) into bytecode (.pyc) then the interpreter executes that bytecode to generate the output.

- ### Interpreted:
    - An interpreter executes the code line by line, and if it encounters any error it stops the execution at that point.

- ### Interactive:
    - Python can interact directly with users through interactive shells and it can interface with other programming languages using various libraries and modules.

- ### Dynamically typed:
    - Python automatically determines the data type during the runtime, so we don't need to declare it explicitly.

- ### Automatic Memory Managemet:
    - It automatically handles memory allocation and allocates single memory block if data/volume is same.

- ### Garabage Collector:
    - It automatically removes unused memory block.

- ### Loosely Coupled:
    - Its components, modules and functions are independent of each other.

- ### Case Sensitive:
    - Python treats uppercase and lowercase differently, so we must call the object exactly the way it is namd.

- ### OOPs:
    - In Python, Object Oriented Programming is a programming paradigm that uses class and object. It aims to use real world entities like inheretance, encapsulation, polymorphism etc in the programming. The main aim of OOPs is to bind the data and function that work on it together as a single unit so that no other part of the code can have access to it.

## Variables:
A variable is a named location that is used to store the data in the memory. It is helpful to think variable as a container that holds the data that can be changed later in Python. Python don't actually assigns value to the variable instead it gives referece of the value of the variable.

### Rules of variable name:
- A variable name can start with alphabet and underscore.
- A variable name cannot start with number.
- A variable name must contain alpha numeric characters and underscore. (A-Z, a-z,0-9)
- A variable name is case sensitive. (Age, age, AGE all are different)

```python
X = 5
print(X)
```

## Python Operators

Operators are used to perform operations on variables and values.

- ### Arithmetic Operator
    - Arithmetic operators are used with numeric values to perform common mathemtical operations. Ex. +, -, *, /, %, //, **

```python
x = 5
y = 2
z = -13

print(x + y) # Addition
print(x - y) # Subtraction
print(x * y) # Multiplication
print(x / y) # Division
print(x // y) # Floor division
print(x % y) # MOdulus
print(abs(z)) #Absoulte value of a number
print(x ** y) # Raising to power
print(round(2.1)) # Rounds a number
print(bin(37)) # Binary version of a number
print(hex(37)) #hexadecimal version of a number
```

- ### Assignment Operator
    - Assignment Operators are used to assign values to variables. Ex. +=,-=, *=, /=, %=, **=

```python
x = 21
print(x)

x += 3
print(x)

x -= 3
print(x)

x *= 3
print(x)

x /= 3
print(x)

x //= 3
print(x)

x %=3
print(x)
```

- ### Comparison Operator
    - Comparison Operator is used to compare two values. Ex. ==, !=, >, <, >=, <=

```python
i = 3
j = 7

print(i > j) # Greater than
print(i < j) # less than
print(i >=j) # Greater than or eual
print(i <=j) # less than or equal
print(i ==j) # Equality
print(i !=j) # Inequality
```

- ### Identity Operator 
    - Identity operator are used to check the object, not if they are equal but to check if they are the same object and have same memory allocation. Ex. is, is not

- ### Membership Operator
    - Membership operator is used to check if a sequence is presented in the object. Ex. is, is not

- ### Bitwise Operator
    - Bitwise operator is used to compare binary numbers. Ex. &, |, ^, ~, <<, >>
