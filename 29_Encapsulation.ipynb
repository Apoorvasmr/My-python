# Python Encapsulation
Encapsulation means keeping data and the methods that work with it together in a class. A class can offer methods for using or changing data in a controlled way.

In Python, underscores are conventions that communicate how names should be used. They do not provide strict privacy.

## Private Properties
A property name beginning with two underscores, such as `__balance`, is name-mangled by Python. This helps avoid accidental name conflicts, but it does not make the property truly private.

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance

account = Account(100)
print("Account created.")
```

## Get a Private Property Value
A class method can read a property that begins with two underscores. This provides a clear way to ask the object for its value.

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = Account(100)
print(account.get_balance())
```

## Set a Private Property Value
A method can change the property after checking the new value. This helps keep invalid values out of the object.

```python
class Account:
    def __init__(self, balance):
        self.__balance = balance

    def set_balance(self, amount):
        if amount < 0:
            raise ValueError("Balance cannot be negative.")
        self.__balance = amount

    def get_balance(self):
        return self.__balance

account = Account(100)
account.set_balance(150)
print(account.get_balance())
```

## Why Use Encapsulation?
Methods can check changes before saving them. This helps protect an object's data from invalid values and keeps the rules in one place.

```python
class Thermometer:
    def __init__(self, temperature):
        self.__temperature = temperature

    def set_temperature(self, temperature):
        if temperature < -273.15:
            raise ValueError("Temperature is below absolute zero.")
        self.__temperature = temperature

    def get_temperature(self):
        return self.__temperature

thermometer = Thermometer(20)
thermometer.set_temperature(22)
print(thermometer.get_temperature())
```

## Protected Properties
A name beginning with one underscore, such as `_level`, is a convention that says the property is intended for use inside the class and its subclasses. Python still allows access to it.

```python
class Animal:
    def __init__(self, name):
        self._name = name

class Dog(Animal):
    def introduce(self):
        print("I am", self._name)

dog = Dog("Pip")
dog.introduce()
```

## Private Methods
A method whose name begins with two underscores is name-mangled too. It is usually called by another method in the same class, not used as part of the class's public interface.

```python
class Greeter:
    def __make_message(self, name):
        return "Hello, " + name

    def greet(self, name):
        print(self.__make_message(name))

Greeter().greet("Ari")
```

## Name Mangling
Python changes a name like `__value` to include the class name, helping prevent accidental clashes in subclasses. This is called name mangling. It is not a security feature or a way to make data inaccessible.

```python
class Sample:
    def __init__(self):
        self.__value = 7

sample = Sample()
print(sample._Sample__value)  # Shows the name-mangled form; usually avoid this.
```
