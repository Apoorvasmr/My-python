# Python File Handling
File handling lets a program read data from files and write data to files. This chapter uses temporary folders for examples so your existing files are not changed.

## Opening a File
Use `open()` with a file path and a mode. The common modes are `r` to read, `w` to write (replace existing content), `a` to add to the end, and `x` to create a new file. Text mode is the default.

### Syntax
The basic form is `open(path, mode)`. For text files, you can also set an encoding, such as `encoding="utf-8"`.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as folder:
    path = Path(folder) / "notes.txt"
    path.write_text("Hello from a file!", encoding="utf-8")

    with open(path, "r", encoding="utf-8") as file:
        print(file.read())
```

## Using the `with` Statement
The `with` statement closes a file automatically when its block ends, even if an error happens. It is the recommended way to work with files.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "message.txt"
    path.write_text("Files are closed automatically.", encoding="utf-8")

    with open(path, encoding="utf-8") as file:
        print(file.read())
```

## Closing Files
If you open a file without `with`, call `close()` when you finish. A file left open may keep using system resources or may not save all changes right away. Prefer `with` for normal use.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "message.txt"
    path.write_text("Close the file when done.", encoding="utf-8")

    file = open(path, encoding="utf-8")
    try:
        print(file.read())
    finally:
        file.close()
```

## Reading a File
Open a text file in read mode (`r`) and call `read()` to get its contents as a string.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "story.txt"
    path.write_text("A small story.", encoding="utf-8")

    with open(path, "r", encoding="utf-8") as file:
        contents = file.read()
    print(contents)
```

## Reading Only Part of a File
Give `read()` a number to read at most that many characters. A later call continues from the current position in the file.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "words.txt"
    path.write_text("Python makes useful tools.", encoding="utf-8")

    with open(path, encoding="utf-8") as file:
        print(file.read(6))
        print(file.read(5))
```

## Reading Lines
Use `readline()` to read one line at a time. You can also loop over the file to process each line.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "list.txt"
    path.write_text("apples\npears\nplums\n", encoding="utf-8")

    with open(path, encoding="utf-8") as file:
        for line in file:
            print(line.strip())
```

## Writing to an Existing File
Mode `a` adds text to the end of a file. Mode `w` replaces the file's contents, so use it carefully. These examples use temporary files.

### Add to the End of a File
Open the file with mode `a` to keep its current content and add more.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "log.txt"
    path.write_text("First line\n", encoding="utf-8")

    with open(path, "a", encoding="utf-8") as file:
        file.write("Added line\n")

    print(path.read_text(encoding="utf-8"))
```

### Replace Existing Content
Opening a file with mode `w` clears its old content before writing. Make sure you have the right file before using this mode.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "draft.txt"
    path.write_text("Old text", encoding="utf-8")

    with open(path, "w", encoding="utf-8") as file:
        file.write("New text")

    print(path.read_text(encoding="utf-8"))
```

## Creating a New File
Mode `x` creates a new file and raises `FileExistsError` if the file is already there. Mode `w` also creates a file if it does not exist, but replaces it if it does.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "new.txt"

    with open(path, "x", encoding="utf-8") as file:
        file.write("This is a new file.")

    print(path.read_text(encoding="utf-8"))
```

## Deleting a File
Use `unlink()` to remove a file. Check that the path is the file you intend to remove. This example only deletes a file inside its temporary folder.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "temporary.txt"
    path.write_text("Temporary data", encoding="utf-8")

    path.unlink()
    print("File exists:", path.exists())
```

### Check if a File Exists
Use `exists()` before trying to remove a path. This avoids an error when the file is not there.

```python
with TemporaryDirectory() as folder:
    path = Path(folder) / "temporary.txt"
    path.write_text("Temporary data", encoding="utf-8")

    if path.exists():
        path.unlink()
        print("File removed.")
    else:
        print("File not found.")
```

## Deleting a Folder
Use `rmdir()` to remove an empty folder. It raises an error if the folder contains files. Be careful to check the folder path before removing it. This example removes only an empty folder inside a temporary directory.

```python
with TemporaryDirectory() as folder:
    empty_folder = Path(folder) / "empty"
    empty_folder.mkdir()

    empty_folder.rmdir()
    print("Folder exists:", empty_folder.exists())
```
