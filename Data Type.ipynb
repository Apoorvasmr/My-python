## Data Type
Data type is the categorization or classification of the data items. It represents the kind of value that tells what operation can be used to perform on a particular data. Since everything is an object in the python programming language, data type is the class and variables are the instances of this classes.

### Primitive Data Structure
They are the most primitive or basic form of data structure. They are the building block for data manipulation and contain pure simple values of data. Example- Integer, float , boolean , string

### Non-primitive Data Structure
Non-primitive data structure are the most sophisticated member of the family. They don't just store values but rather collection of values in various format. Example- Array, files, lists

Python has several inbuilt data types, some of them are-
- Numeric data type
- Text sequence data type
- Sequence data type
- Mapping data type
- Boolean data type
- Set data type

The type() function is used to check the data type of something in Python.

## Numeric Data Type
In python, numeric data type refers to the data which has numeric value. Numeric value can be integer, float or complex number.

- ### Integer
This value is represented by int class. It contains positive and negative whole number without decimals.

- ### Float
This value represents float number. It contains real number with floating point representation. It has decimal point.

- ### Complex Number
This value is represented by complex class. It has real + imaginary part (j).

```python
x = 7 # integer
y = 3.14 # decimal
z = 5 - 2j #complex
print(type(x))
print(type(y))
print(type(z))
```

## Text Sequence Data Type
- String (`str`) is a text data type that stores characters.
- Example: "hello", "Python"

## Sequence Data Type
- Sequence data types store ordered collections of items.
- Examples: list, tuple, range

## Mapping Data Type
- Mapping data types store data in key-value pairs.
- Example: dictionary (`dict`)

## Boolean Data Type
- Boolean (`bool`) stores only two values: `True` or `False`.

## Set Data Type
- Set data types store unordered values without duplicate elements.
- Example: set, frozenset
