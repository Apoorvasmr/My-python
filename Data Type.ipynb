## Data Type
Data type is the categorization or classification of the data items. It represents the kind of value that tells what operation can be used to perform on a particular data. Since everything is an object in the python programming language, data type is the class and variables are the instances of this classes.

### Primitive Data Structure
They are the most primitive or basic form of data structure. They are the building block for data manipulation and contain pure simple values of data. Integer, float , boolean , string

### Non-primitive Data Structure
Non-primitive data structure are the most sophisticated member of the family. They don't just store values but rather collection of values in various format. Array, files, lists
