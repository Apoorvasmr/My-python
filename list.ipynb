## Python Lists
A list is a collection that stores multiple items in one variable. Lists are ordered, changeable (mutable), and allow duplicate values. Each item has a position called an index, and indexing starts at 0. A list can contain items of the same or different data types.


```python
fruits = ["apple", "banana", "cherry", "apple"]
print(fruits)
```

## Lists are ordered
A list is ordered, so its items keep their positions. Adding an item with `append()` places it at the end, unless an operation explicitly rearranges the list.


```python
colors = ["red", "green", "blue"]
colors.append("yellow")
print(colors[0])
print(colors)
```

## Lists are changeable
Lists are mutable, which means you can replace, add, or remove items after creating the list. The list itself is updated by these operations.


```python
colors = ["red", "green", "blue"]
colors[1] = "purple"
print(colors)
```

## Lists allow duplicate values
A list can contain the same value more than once. Each occurrence still has its own index and contributes to the list length.


```python
fruits = ["apple", "banana", "cherry", "apple", "cherry"]
print(fruits)
print(len(fruits))
```

## Creating a list
Use square brackets `[]` to create a list. Separate each item with a comma. An empty list has no items.


```python
fruits = ["apple", "banana", "cherry"]
empty_list = []
print(fruits)
print(empty_list)
```

## List length
The built-in `len()` function returns the number of items in a list. Duplicate values are counted as separate items.


```python
fruits = ["apple", "banana", "apple"]
print(len(fruits))
```

## List item data types
List items can be strings, numbers, booleans, or other Python values. A single list can also contain different data types.


```python
strings = ["apple", "banana"]
numbers = [1, 2, 3]
flags = [True, False]
mixed = ["age", 20, True]
print(strings)
print(numbers)
print(flags)
print(mixed)
```

## List type
The built-in `type()` function shows an object's type. A list has the type `list`.


```python
fruits = ["apple", "banana", "cherry"]
print(type(fruits))
```

## The list() constructor
The `list()` constructor creates a list from an iterable, such as a tuple or a string. When creating a list from a tuple, pass the tuple as its argument.


```python
colors = list(("red", "green", "blue"))
print(colors)
```

## Accessing list items
Use an item's index inside square brackets to access it. Index `0` is the first item; negative indexes count backward from the end, where `-1` is the last item.


```python
colors = ["red", "green", "blue"]
print(colors[0])
print(colors[1])
print(colors[-1])
print(colors[-2])
```

## Range of list indexes
A slice uses `list[start:stop]` to make a new list. The start index is included and the stop index is excluded. You may leave out either bound; negative indexes count from the end.


```python
fruits = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(fruits[2:5])
print(fruits[:4])
print(fruits[2:])
print(fruits[-4:-1])
```

## Check whether an item exists
Use the `in` operator to check whether a value appears in a list. The expression returns `True` or `False`.


```python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
print("mango" in fruits)
```

## Changing a list item
Assign a new value to an item by using its index. This changes the existing list in place.


```python
fruits = ["apple", "banana", "cherry"]
fruits[1] = "blackcurrant"
print(fruits)
```

## Changing a range of list items
Assign a new list to a slice to replace several items. If the replacement has a different number of items, the list length changes.


```python
fruits = ["apple", "banana", "cherry", "orange"]
fruits[1:3] = ["blackcurrant", "watermelon"]
print(fruits)

fruits[1:3] = ["pear"]
print(fruits)
```

## Adding items to a list
New items can be added at the end, at a chosen position, or from another iterable. The examples below show each operation separately.


```python
fruits = ["apple", "banana"]
fruits.append("cherry")
print(fruits)
```

## append()
The `append()` method adds one item to the end of the list. It changes the original list and returns `None`.


```python
fruits = ["apple", "banana"]
result = fruits.append("cherry")
print(fruits)
print(result)
```

## insert()
The `insert()` method adds one item at a specified index. Existing items at that position and after it move to the right.


```python
fruits = ["apple", "cherry"]
fruits.insert(1, "banana")
print(fruits)
```

## extend()
The `extend()` method adds each item from an iterable to the end of the list. Unlike `append()`, it adds iterable items individually.


```python
fruits = ["apple", "banana"]
more_fruits = ("cherry", "orange")
fruits.extend(more_fruits)
print(fruits)
```

## Removing items from a list
A list item can be removed by its value, by its index, with `del`, or by clearing all items. Removing an item changes the original list.


```python
fruits = ["apple", "banana", "cherry"]
del fruits[1]
print(fruits)
```

## remove()
The `remove()` method deletes the first item whose value matches the argument. It raises `ValueError` if the value is not present.


```python
fruits = ["apple", "banana", "cherry", "banana"]
fruits.remove("banana")
print(fruits)
```

## pop()
The `pop()` method removes and returns the item at an index. If no index is supplied, it removes and returns the last item.


```python
fruits = ["apple", "banana", "cherry"]
removed_item = fruits.pop(1)
print(removed_item)
print(fruits)
```

## del keyword
The `del` keyword removes an item by index. It can also delete the entire list variable, after which that variable cannot be used until it is created again.


```python
fruits = ["apple", "banana", "cherry"]
del fruits[0]
print(fruits)
```

## clear()
The `clear()` method removes all items but keeps the list variable, leaving it as an empty list.


```python
fruits = ["apple", "banana", "cherry"]
fruits.clear()
print(fruits)
```

## Loop through a list
A `for` loop visits each item in order, one at a time. Use the item variable directly when you do not need its index.


```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

## Loop using index numbers
Use `range(len(list_name))` to loop over index numbers. This is useful when the index itself is needed.


```python
fruits = ["apple", "banana", "cherry"]
for index in range(len(fruits)):
    print(index, fruits[index])
```

## Loop with while
A `while` loop can visit list items by keeping an index. Increase the index each time so the loop eventually stops.


```python
fruits = ["apple", "banana", "cherry"]
index = 0
while index < len(fruits):
    print(fruits[index])
    index += 1
```

## List comprehension
A list comprehension creates a new list from an iterable in one expression. It can include an optional condition to filter items.


```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
new_list = [fruit for fruit in fruits if "a" in fruit]
print(new_list)
```

## List comprehension syntax
The general form is `[expression for item in iterable if condition]`. The `if condition` part is optional; the expression determines each item in the new list.


```python
numbers = [number for number in range(6)]
even_numbers = [number for number in range(6) if number % 2 == 0]
print(numbers)
print(even_numbers)
```

## Transforming values in a comprehension
The expression at the start of a comprehension can transform each selected item as the new list is created.


```python
fruits = ["apple", "banana", "cherry"]
uppercase_fruits = [fruit.upper() for fruit in fruits]
print(uppercase_fruits)
```

## Conditional expressions in a comprehension
An `if ... else` inside the expression chooses a new value for each item. Unlike a trailing filter condition, it keeps every item and transforms some of them.


```python
fruits = ["apple", "banana", "cherry"]
updated_fruits = ["orange" if fruit == "banana" else fruit for fruit in fruits]
print(updated_fruits)
```

## sort(): ascending order
The `sort()` method rearranges a list in place. By default, values are sorted in ascending alphanumeric order; values in one list should be mutually comparable.


```python
numbers = [100, 50, 65, 82, 23]
numbers.sort()
print(numbers)
```

## Sorting in descending order
Pass `reverse=True` to `sort()` to arrange the list from highest to lowest, or in reverse alphabetical order for strings.


```python
numbers = [100, 50, 65, 82, 23]
numbers.sort(reverse=True)
print(numbers)
```

## Custom sort key
Use the `key` parameter to tell `sort()` which value to use for comparison. The method still sorts the original list in place.


```python
numbers = [100, 50, 65, 82, 23]
numbers.sort(key=lambda number: abs(number - 50))
print(numbers)
```

## Case-insensitive sorting
String sorting is case-sensitive by default. Use `key=str.lower` to compare strings without considering uppercase or lowercase differences.


```python
fruits = ["banana", "Orange", "kiwi", "cherry"]
fruits.sort(key=str.lower)
print(fruits)
```

## reverse()
The `reverse()` method reverses the current order of items. It does not sort them; it simply swaps the order in place.


```python
fruits = ["banana", "orange", "kiwi", "cherry"]
fruits.reverse()
print(fruits)
```

## Copying a list versus assigning it
Writing `second = first` does not make a separate list; both names refer to the same list. Changes through either name are visible through both.


```python
first = ["red", "green"]
second = first
second.append("blue")
print(first)
print(second)
```

## copy()
The `copy()` method creates a new shallow copy of a list. For a list of simple values, changing the copied list does not change the original list.


```python
original = ["red", "green"]
copy_of_original = original.copy()
copy_of_original.append("blue")
print(original)
print(copy_of_original)
```

## Other ways to copy a list
`list(existing_list)` and `existing_list[:]` each create a shallow copy of a list, just like `copy()`.


```python
original = ["red", "green", "blue"]
copy_with_list = list(original)
copy_with_slice = original[:]
print(copy_with_list)
print(copy_with_slice)
```

## Joining lists with +
The `+` operator combines two lists and returns a new list. It does not change either input list.


```python
letters = ["a", "b", "c"]
numbers = [1, 2, 3]
combined = letters + numbers
print(combined)
print(letters)
```

## Joining lists with extend() or a loop
`extend()` adds items from another list to the first list. A `for` loop with `append()` can add the same items one at a time.


```python
first = ["a", "b"]
second = ["c", "d"]
joined_with_extend = first.copy()
joined_with_extend.extend(second)

joined_with_loop = first.copy()
for item in second:
    joined_with_loop.append(item)

print(joined_with_extend)
print(joined_with_loop)
```

## count()
The `count()` method returns how many times a specified value appears in the list.


```python
numbers = [1, 2, 2, 3, 2]
print(numbers.count(2))
```

## index()
The `index()` method returns the position of the first matching value. It raises `ValueError` if the value is not found.


```python
fruits = ["apple", "banana", "cherry", "banana"]
print(fruits.index("banana"))
```
