## Python Loops
A loop is used to repeat a block of code multiple times. In Python, the two main loop types are `while` and `for`.

Loops help us avoid writing the same code again and again.

## While loop
A `while` loop runs as long as a condition is `True`.
It is useful when the number of iterations depends on a condition.

```python
i = 1
while i < 6:
    print(i)
    i += 1
```

## While loop with break
The `break` statement stops the loop immediately, even if the condition is still `True`.

```python
i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1
```

## While loop with continue
The `continue` statement skips the remaining code in the current iteration and moves to the next one.

```python
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
```

## While loop else
The `else` block after a `while` loop runs when the loop condition becomes `False`.
This is useful for a message after the loop ends.

```python
i = 1
while i < 6:
    print(i)
    i += 1
else:
    print("i is no longer less than 6")
```

## For loop
A `for` loop is used to iterate over a sequence such as a list, tuple, string, set, or dictionary.
It is usually used when we know what we are looping through.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

## Looping through a string
Strings are also iterable, so a `for` loop can process each character.

```python
for letter in "banana":
    print(letter)
```

## For loop with break
The `break` statement stops the loop before all items are processed.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
    if fruit == "banana":
        break
```

## For loop with continue
The `continue` statement skips the current item and moves to the next one.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    if fruit == "banana":
        continue
    print(fruit)
```

## The range() function
The `range()` function generates a sequence of numbers.
This is very useful in `for` loops when you want to repeat a block a fixed number of times.

```python
for x in range(6):
    print(x)

print("---")
for x in range(2, 8):
    print(x)

print("---")
for x in range(1, 10, 2):
    print(x)
```

## For loop else
The `else` block in a `for` loop executes after the loop finishes normally.
It does not run if the loop ends because of a `break`.

```python
for x in range(3):
    print(x)
else:
    print("Loop finished successfully")
```

## Nested loops
A nested loop is a loop inside another loop.
The inner loop runs completely for every iteration of the outer loop.

```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for a in adj:
    for b in fruits:
        print(a, b)
```

## The pass statement in loops
A loop cannot be empty. If you ever need an empty loop, use `pass` to avoid an error.

```python
for x in [0, 1, 2]:
    pass

print("Loop finished without errors")
```
