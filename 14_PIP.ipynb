# Python PIP
PIP is a tool for installing and managing Python packages. A package adds code that you can use in your own programs.

## What Is PIP?
PIP downloads packages from the Python Package Index (PyPI) and installs them for Python. Run it from a terminal or command prompt.

## What Is a Package?
A package is a collection of Python modules and other files. A module is a Python file that contains code you can use in another program.

## Check if PIP Is Installed
Use `python -m pip --version` in a terminal. Using `python -m pip` runs PIP for the Python interpreter selected by `python`.

```text
python -m pip --version
```

## Install PIP
PIP is included with most current Python installations. If it is missing, try Python's built-in `ensurepip` module. Some Python installations manage packages another way; follow the installation instructions for that Python if this command is unavailable.

```text
python -m ensurepip --upgrade
python -m pip --version
```

## Download a Package
Use `install` followed by the package name. This command installs the `camelcase` package. Run terminal commands in a terminal, not in a Python code cell.

```text
python -m pip install camelcase
```

## Using a Package
After installing a package, import it in your Python program. This example uses `camelcase`; run it after installing that package.

The class is named `CamelCase` inside the `camelcase` module. Use a dot to access it:

````python
import camelcase

tool = camelcase.CamelCase()
print(tool.hump("hello from python"))
````

## Find Packages
Search for packages on [PyPI](https://pypi.org/). Check a package's description, documentation, supported Python versions, and release history before installing it.

## Remove a Package
Use `uninstall` followed by the package name. PIP asks you to confirm before removing it.

```text
python -m pip uninstall camelcase
```

## List Packages
Use `list` to see packages installed for the selected Python interpreter.

```text
python -m pip list
```

Use `show` to see information about one installed package, such as its version and location.

```text
python -m pip show camelcase
```

## Use a Virtual Environment
A virtual environment keeps a project's packages separate from other projects. On Windows, create one with `python -m venv .venv` and activate it with `.venv\Scripts\activate`. Then use `python -m pip` to install packages in that environment.

```text
python -m venv .venv
.venv\Scripts\activate
python -m pip install camelcase
```
