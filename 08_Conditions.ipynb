## Python Conditions
A condition is a test that returns either `True` or `False`. In Python, conditions are often used with `if`, `elif`, and `else` to make decisions in a program.

Python supports comparison operators such as `==`, `!=`, `<`, `<=`, `>`, and `>=`. These help us check whether something is equal, greater, smaller, or different.

```python
a = 33
b = 200

if b > a:
    print("b is greater than a")
```

## Comparison operators
These operators compare values and return a boolean result. A boolean is just `True` or `False`.

```python
x = 10
y = 15

print(x == y)
print(x != y)
print(x < y)
print(x <= y)
print(x > y)
print(x >= y)
```

## How if statements work
The `if` statement checks a condition. If the condition is `True`, the block inside the `if` runs. If it is `False`, the block is skipped.

```python
number = 15

if number > 0:
    print("The number is positive")
```

## Indentation in Python
Python uses indentation to group statements together. The code inside an `if` block must be indented. Without indentation, Python raises an error.

```python
age = 20

if age >= 18:
    print("You are an adult")
    print("You can vote")
    print("You have full legal rights")
```

## The else statement
The `else` block runs when the `if` condition is `False`. It is the fallback option when the first condition does not match.

```python
a = 200
b = 33

if b > a:
    print("b is greater than a")
else:
    print("b is not greater than a")
```

## The elif statement
The `elif` keyword means `else if`. It lets us check multiple conditions in order. Python stops at the first condition that is `True`.

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

## Short hand if
If the block has only one statement, we can write the `if` on a single line. This is called a shorthand `if` statement.

```python
x = 10
if x > 5: print("x is greater than 5")
```

## Short hand if else
The shorthand `if else` form is useful when we want a quick two-way decision.

```python
y = 3
print("Even") if y % 2 == 0 else print("Odd")
```

## Logical operators
The logical operators `and`, `or`, and `not` help us combine conditions. They are useful when we need more than one test in a decision.

```python
temperature = 25
is_sunny = True

if temperature > 20 and is_sunny:
    print("It is a good day for a walk")

if temperature < 10 or is_sunny:
    print("At least one condition is true")

if not is_sunny:
    print("It is not sunny")
else:
    print("It is sunny")
```

## Truthy and falsy values
Python treats some values as `False` in a condition and others as `True`.

Values considered False:
- `0`
- `0.0`
- `""` (empty string)
- `None`
- empty lists, tuples, dictionaries, sets, and other empty collections

Everything else is usually treated as `True`.


```python
values = [0, 0.0, "", None, [], {}, set()]
for v in values:
    print(repr(v), "->", bool(v))

print(bool(5))
print(bool("Hello"))
print(bool([1, 2, 3]))
```

## Multiple statements in one if block
You can write more than one statement inside an `if` block. All statements in the same block must have the same indentation.


```python
age = 20
if age >= 18:
    print("You are eligible")
    print("You can register")
    print("Welcome!")
```

## The pass statement
If you want an empty `if` block, you can use the `pass` statement. It does nothing but prevents Python from raising a syntax error.


```python
marks = 40
if marks >= 50:
    pass
else:
    print("You need more marks")
```

## Nested if statements
A nested `if` is an `if` statement inside another `if` statement. This is used when a decision depends on a prior decision.

```python
age = 20
has_id = True

if age >= 18:
    print("You are old enough")
    if has_id:
        print("You can enter the club")
    else:
        print("You need an ID to enter")
else:
    print("You are too young")
```

## Boolean variables in conditions
A boolean variable can be used directly in an `if` statement. Python treats `True` as true and `False` as false.

```python
is_logged_in = True

if is_logged_in:
    print("Welcome back!")
else:
    print("Please log in")
```

## The match statement
The `match` statement compares a value with patterns in `case` blocks and runs the first matching block. It is useful when choosing between several possible values. `match`/`case` requires Python 3.10 or later.


```python
day = 3

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
```

## The default case
Use `case _:` as a fallback when none of the earlier cases match. The underscore is a wildcard pattern and is usually placed last.


```python
day = 7

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case _:
        print("No weekday case matched")
```

## Match multiple values
Use the `|` pattern to let one `case` match any of several values. This is like checking several alternatives with `or`.


```python
status_code = 404

match status_code:
    case 200 | 201:
        print("Request succeeded")
    case 400 | 401 | 403 | 404:
        print("There is a request or access problem")
    case _:
        print("Other status")
```

## Add a condition to a case
A case can include an `if` guard. The pattern must match first, and then the guard condition must also be `True` for that case to run.


```python
age = 20
has_id = True

match age:
    case value if value >= 18 and has_id:
        print("Entry allowed")
    case value if value >= 18:
        print("An ID is required")
    case _:
        print("Too young")
```
